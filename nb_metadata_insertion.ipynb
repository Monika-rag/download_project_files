```python
import datetime as dt
from decimal import Decimal

table_name = "dbo.md_ingestion_config"

df = spark.table(table_name)
cols = df.columns
rows = df.collect()

def to_sql_value(value):
    if value is None:
        return "NULL"

    if isinstance(value, bool):
        return "true" if value else "false"

    if isinstance(value, int):
        return str(value)

    if isinstance(value, float):
        return str(value)

    if isinstance(value, Decimal):
        return str(value)

    if isinstance(value, dt.datetime):
        formatted_value = value.strftime("%Y-%m-%d %H:%M:%S.%f")
        return f"CAST('{formatted_value}' AS TIMESTAMP)"

    if isinstance(value, dt.date):
        formatted_value = value.strftime("%Y-%m-%d")
        return f"CAST('{formatted_value}' AS DATE)"

    value_str = str(value).replace("'", "''")
    return f"'{value_str}'"

insert_rows = []

for row in rows:
    values = [to_sql_value(row[col]) for col in cols]
    insert_rows.append("(" + ", ".join(values) + ")")

column_list = ", ".join(cols)
joined_rows = ",\n".join(insert_rows)

insert_script = f"""
INSERT INTO {table_name}
(
    {column_list}
)
VALUES
{joined_rows};
"""

print(insert_script)
```
